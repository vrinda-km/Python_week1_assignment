```python
#Q31)
result = []
count = 0
for num in range(1, 10000):
    temp = num
    digits = 0
    while temp > 0:
        digits = digits + 1
        temp = temp // 10
    temp = num
    total = 0
    while temp > 0:
        digit = temp % 10
        total = total + digit ** digits
        temp = temp // 10
    if total == num:
        result.append(num)
        count = count + 1

print("Armstrong numbers:", result)
print("Count:", count)
```

```python
#Q38)
review = input("Enter a review: ").lower()
words = review.split()
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1
sorted_words = []
for word in frequency:
    sorted_words.append(word)
for i in range(len(sorted_words)):
    for j in range(i + 1, len(sorted_words)):
        if frequency[sorted_words[j]] > frequency[sorted_words[i]]:
            temp = sorted_words[i]
            sorted_words[i] = sorted_words[j]
            sorted_words[j] = temp
for word in sorted_words:
    print(word, ":", frequency[word])
most_frequent = sorted_words[0]
print("Most frequent word:", most_frequent)
print("Count:", frequency[most_frequent])
```

```python
#Q39)
states = {
    "Karnataka": "Bengaluru",
    "Kerala": "Thiruvananthapuram",
    "Maharashtra": "Mumbai",
    "Tamil Nadu": "Chennai",
    "Telangana": "Hyderabad"
}

while True:
    print("\n1. Add a new state-capital pair")
    print("2. Update an existing capital")
    print("3. Delete a state")
    print("4. Search a state")
    print("5. Display all states")
    print("6. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        state = input("Enter state name: ")
        capital = input("Enter capital name: ")
        states[state] = capital
        print("State added successfully")

    elif choice == 2:
        state = input("Enter state name: ")
        if state in states:
            capital = input("Enter new capital: ")
            states[state] = capital
            print("Capital updated successfully")
        else:
            print("State not found")
    elif choice == 3:
        state = input("Enter state name: ")
        if state in states:
            del states[state]
            print("State deleted successfully")
        else:
            print("State not found")
    elif choice == 4:
        state = input("Enter state name: ")
        if state in states:
            print("Capital:", states[state])
        else:
            print("Not found")
    elif choice == 5:
        state_list = list(states.keys())

        for i in range(len(state_list)):
            for j in range(i + 1, len(state_list)):
                if state_list[i] > state_list[j]:
                    temp = state_list[i]
                    state_list[i] = state_list[j]
                    state_list[j] = temp
        for state in state_list:
            print(state, ":", states[state])
    elif choice == 6:
        print("Exiting program")
        break
    else:
        print("Invalid choice")
```

```python
# Q40)
section_a = {"Python", "Java", "C++", "SQL"}
section_b = {"Java", "JavaScript", "Python", "Go", "SQL"}
# (a) Languages known by both sections
both = section_a.intersection(section_b)
# (b) Languages known only by Section A
only_a = section_a.difference(section_b)
# (c) Languages known only by Section B
only_b = section_b.difference(section_a)
# (d) Total unique languages
unique = section_a.union(section_b)
# (e) Languages exclusive to one section
exclusive = section_a.symmetric_difference(section_b)
print("Languages known by both:", both)
print("Languages known only by Section A:", only_a)
print("Languages known only by Section B:", only_b)
print("Total unique languages:", unique)
print("Languages exclusive to one section:", exclusive)
```

```python
# Q41)
prices = (3845, 3902, 3878, 3955, 4012, 3998, 4055)
days = ("Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday")
maximum = prices[0]
minimum = prices[0]
max_day = days[0]
min_day = days[0]
total = 0
for i in range(7):
    total = total + prices[i]
    if prices[i] > maximum:
        maximum = prices[i]
        max_day = days[i]
    if prices[i] < minimum:
        minimum = prices[i]
        min_day = days[i]
average = total / 7
biggest_gain = prices[1] - prices[0]
gain_day = days[0] + " to " + days[1]
volatility = 0
for i in range(1, 7):
    change = prices[i] - prices[i - 1]
    if change > biggest_gain:
        biggest_gain = change
        gain_day = days[i - 1] + " to " + days[i]
    if change < 0:
        change = -change
    volatility = volatility + change
print("Maximum price:", maximum, "on", max_day)
print("Minimum price:", minimum, "on", min_day)
print("Average price:", average)
print("Biggest single-day gain:", biggest_gain, "between", gain_day)
print("Total volatility:", volatility)
```

```python
# Q46)
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    if gst_inclusive == False:
        base_amount = amount
        gst_amount = base_amount * gst_rate / 100
        total_amount = base_amount + gst_amount
    else:
        total_amount = amount
        base_amount = amount / (1 + gst_rate / 100)
        gst_amount = total_amount - base_amount
    return (round(base_amount, 2), round(gst_amount, 2), round(total_amount, 2))
print(calculate_gst(1000))
print(calculate_gst(1180, gst_inclusive=True))
print(calculate_gst(500, gst_rate=5))
```

```python
# Q47)
def power(base, exp):
    if exp == 0:
        return 1
    elif exp < 0:
        return 1 / power(base, -exp)
    else:
        return base * power(base, exp - 1)
print(power(2, 10))
print(power(5, 0))
print(power(3, -2))
```

```python
# Q49)
import numpy as np
sales = np.array([12000, 18500, 14200, 22100, 19800, 35200, 28400])
days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])
# (a) Total weekly sales
total = np.sum(sales)
# (b) Daily average
average = np.mean(sales)
# (c) Day with maximum sales
max_index = np.argmax(sales)
max_day = days[max_index]
# (d) Number of days with sales above average
above_average = np.sum(sales > average)
# (e) Percentage growth from Monday to Sunday
growth = ((sales[6] - sales[0]) / sales[0]) * 100
# (f) Day-on-day percentage changes
changes = (np.diff(sales) / sales[:-1]) * 100
print("Total weekly sales: ₹", total)
print("Daily average: ₹", average)
print("Day with maximum sales:", max_day)
print("Days above average:", above_average)
print("Monday to Sunday growth:", growth, "%")
print("Day-on-day percentage changes:", changes)
```
